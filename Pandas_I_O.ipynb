# **Lendo dadaos SQL**

```python
import sqlalchemy
import pandas as pd
```

```python
from sqlalchemy import create_engine, MetaData, Table, inspect
```

```python
engine = create_engine('sqlite:///:memory:')
```

```python
url = 'https://raw.githubusercontent.com/alura-cursos/Pandas/main/clientes_banco.csv'
```

```python
dados_sql = pd.read_csv(url)
```

```python
dados_sql.head()
```

```python
dados_sql.to_sql("clientes",engine,index=False)
```

```python
inspector = inspect(engine)
```

```python
print(inspector.get_table_names())
```

## **Fazendo consulta SQL**

```python
query = 'SELECT * FROM clientes WHERE Categoria_de_renda="Empregado" '
```

```python
empregados = pd.read_sql(query,engine)
```

```python
empregados
```

**salvando consulta em uma nova tabela no banco**

```python
empregados.to_sql("empregados",con=engine,index=False)
```

**consultando tabela espeficica**

```python
pd.read_sql_table("empregados",engine)
```

```python
pd.read_sql_table("empregados",engine,columns=['ID_Cliente','Idade','Grau_escolaridade','Tamanho_familia'])
```

**Atualizando Tabelas**

```python
query = 'SELECT * FROM clientes'

pd.read_sql(query,engine)
```

```python
from sqlalchemy import text

query = "DELETE FROM clientes WHERE ID_Cliente=5008804"

with engine.begin() as conn:
    conn.execute(text(query))
```

```python
query = "UPDATE clientes SET Grau_escolaridade='Ensino superior' WHERE ID_Cliente=5008808"
with engine.begin() as conn:
  conn.execute(text(query))
```
